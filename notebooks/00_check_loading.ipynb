```python
import numpy as np
import sys
import os
os.environ['PTNOLATEX'] = '1'
sys.path.append("/Users/ivanzait/Documents/Documents_LM4500/Codes/analysator")
import pytools as pt
from pathlib import Path
sys.path.append(str(Path.cwd().parent))
from src.vdf_helpers import coord_re_to_m

RE = 6371.0e3

path = "/Users/ivanzait/Downloads/bulk.0003408.vlsv"
f = pt.vlsvfile.VlsvReader(path)


coords_re = [-12, 0, 0]
cellid = f.get_cellid(coord_re_to_m(coords_re))

rho = f.read_variable('rho', cellid)
rho_v = f.read_variable('rho_v', cellid)
v = rho_v/rho

B = f.read_variable( 'B' )
#B = f.read_variable( 'B' )
xmin,ymin,zmin,xmax,ymax,zmax=f.get_fsgrid_mesh_extent()
xlen,ylen,zlen =  f.get_fsgrid_mesh_size()
B = np.reshape(B, ( int(xlen),int(ylen),int(zlen),3), order='F')
dx = (xmax - xmin)/xlen
dz = (zmax - zmin)/zlen
xind = int( (coords_re[0]*RE-xmin) / dx)
zind = int( (coords_re[2]*RE-zmin) / dz)
Bval = B[xind,0,zind,:]
print("Bval = ", Bval)
print("v = ", v)


```
